Step 2 - Read the Data

```python
import pandas as pd

df = pd.read_csv('Students_Records.csv')
df.head()
```

Step 3 - Process the Data

```python
import pandas as pd
import numpy as np

subjects = ["Maths","Physics","Chemistry"]

df["Total"] = df[subjects].sum(axis=1)
df["Average"] = df[subjects].mean(axis=1)

grades = []

for average in df["Average"]:
    if average >= 90:
        grade = "A+"
    elif average >= 80:
        grade = "A"
    elif average >= 70:
        grade = "B"
    elif average >= 60:
        grade = "C"
    elif average >= 50:
        grade = "D"
    else:
        grade = "F"

    grades.append(grade)

df["Grade"] = grades

pass_fail = []

for i in range(len(df)):
    if (df.loc[i, "Maths"] >= 40 and
        df.loc[i, "Physics"] >= 40 and
        df.loc[i, "Chemistry"] >= 40 and
        df.loc[i, "Attendance"] >= 75):
        pass_fail.append("Pass")
    else:
        pass_fail.append("Fail")

df["Pass_Fail"] = pass_fail

df[["Student_ID","Name","Total","Average","Grade","Pass_Fail"]]
```

Step 4 - Perform Analysis

```python
highest_student = df.loc[df["Average"].idxmax(),"Name"]
print("Top Student is ",highest_student)

highest_average = df["Average"].max()
print("Highest Average :" ,highest_average)

subject_averages = df[subjects].mean()
best_subject = subject_averages.idxmax()
best_average = subject_averages.max()

print("All subjects averages :\n",subject_averages)
print("Best performance Subject is :",best_subject)
print("The best average is :",best_average)

passed_students = (df["Pass_Fail"] == "Pass").sum()
failed_students = (df["Pass_Fail"] == "Fail").sum()

print(f"Passed Students: {passed_students} | Failed Students: {failed_students}")
```
